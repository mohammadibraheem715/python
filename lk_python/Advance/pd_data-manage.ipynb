```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset("lukebarousse/data_jobs")
df=dataset["train"].to_pandas()

df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
df.sample(10, random_state=42)
```

```python
df.pivot_table(values="salary_year_avg", index="job_title_short", columns="job_country", aggfunc="median")
```

```python
top_countries = df.job_country.value_counts().head().index
top_countries
```

```python
df_cou_sal = df.pivot_table(values="salary_year_avg",columns="job_title_short",index="job_country",aggfunc="median")
```

```python
df_cou_sal = df_cou_sal.loc[top_countries]
```

```python
job_titles = ["Data Analyst","Data Scientist", "Data Engineer"]
df_cou_sal = df_cou_sal[job_titles]
```

```python
df_cou_sal.plot(kind="bar") 
plt.xlabel(" ")
plt.show()
```
