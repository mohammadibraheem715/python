```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

dataset = load_dataset("lukebarousse/data_jobs")
df=dataset["train"].to_pandas()

df["job_posted_date"] = pd.to_datetime(df["job_posted_date"])
```

```python
df.loc[:10,"salary_year_avg":"salary_hour_avg"]
```

```python
med_year = df.salary_year_avg.median()
```

```python
med_hour = df.salary_hour_avg.median()
```

```python
df_filled = df 

df_filled.salary_year_avg = df_filled.salary_year_avg.fillna(med_year)
df_filled.salary_hour_avg = df_filled.salary_hour_avg.fillna(med_hour)
```

```python
df_filled.loc[:10,"salary_year_avg":"salary_hour_avg" ]
```

```python
df_un = df_filled
```

```python
df_un = df_un.drop_duplicates()
```

```python
df_un = df_un.drop_duplicates(subset=["job_title" ,"company_name"])
```
